```python
#Cody Pace
#WGU D598 Task 2
#August 2025

#Import libraries
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.ticker as ticker
import seaborn as sb
sb.set_style('whitegrid')

from IPython.display import display

#Import the Excel file as a data frame
df_input = pd.read_excel("D598 Data Set.xlsx")

#Identify duplicate records, provide count, Business ID for later triage, then remove the duplicates
df_duplicates = df_input[df_input.duplicated()]
duplicates_count = len(df_duplicates)
print(f'Number of duplicate records: {duplicates_count}')

#Provide the Business ID for the duplicate record for easy review.
if len(df_duplicates) > 0:
    print(df_duplicates[['Business ID']])
    df_input = df_input.drop_duplicates()
    print(f'{duplicates_count} duplicate records removed\n')
else:
    print('No duplicates were found in the file.\n')

#Group all rows by Business State
df_state_cols = [ 'Total Long-term Debt', 'Total Equity', 'Debt to Equity', 
                 'Total Liabilities', 'Total Revenue', 'Profit Margin']
df_state_stats = df_input.groupby('Business State')

#Descriptive statistics (mean, median, min, and max)
#for Total Long-term Debt, Total Equity, Debt to Equity,
#Total Liabilities, Total Revenue, Profit Margin
desc_stats = ['mean', 'median', 'min', 'max']
df_desc_stats = df_state_stats[df_state_cols].agg(desc_stats)
df_desc_stats.columns = ['_'.join([col, stat]) for col, stat in df_desc_stats.columns]
print('Descriptive Statistics by State:')
display(df_desc_stats)

#Filter df_desc_stats to identify all businesses with debt-to-equity (dbteq) ratios that are negative
df_negative_dbteq = df_input[df_input['Debt to Equity'] < 0].sort_values(by = 'Debt to Equity')
print('Businesses With A Negative Debt To Equity Ratio')
df_negative_dbteq.reset_index(drop=True, inplace=True)
display(df_negative_dbteq[['Business ID', 'Business State', 'Debt to Equity']])
print(f'There are {len(df_negative_dbteq)} businesses with a negative Debt to Equity Ratio\n')

#Find the Debt-to-income ratio for every business and store in a new data frame.
#Debt-to-income ratio = long-term debt / revenue

#Find businesses with a Total Revenue = $0
df_zero_revenue = df_input[df_input['Total Revenue'] == 0]

#Warn the user if records are found with $0 Total Revenue. Provide Business ID for triage.
if not df_zero_revenue.empty:
    print('WARNING: The following businesses have a Total Revenue = $0 and will be skipped:')
    display(df_zero_revenue[['Business ID']])

#Filter records out
df_filtered_rev = df_input[df_input['Total Revenue'] != 0]
df_input = df_filtered_rev.copy()

#Calculate Debt-to-Income Ratio
df_debt_income_ratio = df_filtered_rev[['Business ID', 'Total Long-term Debt', 'Total Revenue']].copy()
df_debt_income_ratio['Debt-to-Income Ratio'] = df_debt_income_ratio['Total Long-term Debt'] / df_debt_income_ratio['Total Revenue']
df_debt_income_ratio = df_debt_income_ratio.sort_values(by = 'Debt-to-Income Ratio', ascending=False)
print('Debt-to-income ratio for each business')
df_debt_income_ratio.reset_index(drop=True, inplace=True)
display(df_debt_income_ratio)

#Concatenate the debt-to-income ratio data frame with the original data frame
df_final = pd.merge(df_input, df_debt_income_ratio[['Business ID', 'Debt-to-Income Ratio']], on='Business ID', how='left')
print('Final Output')
df_final.reset_index(drop=True, inplace=True)
display(df_final)


#Visualization 1: Bar graph - Top 25 States by Median Total Revenue
#Copy and sort median revenue data
df_viz1 = df_desc_stats[['Total Revenue_median']].copy()
df_viz1 = df_viz1.sort_values(by='Total Revenue_median', ascending=False).head(25)

#Create the bar plot
ax = df_viz1.plot(
    kind='barh',
    legend=False,
    title='Top 25 States by Median Total Revenue',
    figsize=(12, 8),
    color='skyblue'
)

#Formatting
ax.set_xlabel('Median Total Revenue (USD, billions)')
ax.set_ylabel('Business State')
ax.xaxis.set_major_formatter(
    ticker.FuncFormatter(lambda x, pos: (f"{x/1e9:.2f}".rstrip('0').rstrip('.')) + "B"))
ax.grid(True, axis='x', linestyle='--', linewidth=0.5, alpha=0.7)

plt.tight_layout()
plt.savefig("visualization1.png", dpi=300, bbox_inches='tight')
plt.show()


#Visualization 2: Scatter Plot - Profit Margin vs Total Revenue
ax = sb.regplot(
    x='Total Revenue',
    y='Profit Margin',
    data=df_input[df_input['Total Revenue'] > 0],
    scatter=True,
    ci=None,
    line_kws={'color': 'red'},
    scatter_kws={'alpha': 0.5, 's':30},
    lowess=True
)

def money_label(v, pos):
    if v >= 1_000_000_000:
        return f'{v/1_000_000_000:.1f}B'
    elif v >= 1_000_000:
        return f'{v/1_000_000:.1f}M'
    elif v >= 1_000:
        return f'{v/1_000:.0f}K'
    else:
        return f'{v:.0f}'

ax.set_xscale('log')
ax.xaxis.set_major_locator(ticker.LogLocator(base=10.0, numticks=12))
ax.xaxis.set_minor_locator(ticker.LogLocator(base=10.0, subs=(2, 5), numticks=12))
ax.xaxis.set_major_formatter(ticker.FuncFormatter(money_label))
ax.yaxis.set_major_formatter(ticker.PercentFormatter(xmax=100))
ax.collections[0].set_sizes([20])
ax.set_ylim(-25, 25)

#Formatting
plt.title('Do Businesses With More Revenue Have Higher Profit Margins?', fontsize=14)
plt.xlabel('Total Revenue (USD, log scale)', fontsize=12)
plt.ylabel('Profit Margin %', fontsize=12)
plt.grid(True, which='both', linestyle='--', linewidth=0.5, alpha=0.7)

plt.tight_layout()
plt.savefig("visualization2.png", dpi=300, bbox_inches='tight')
plt.show()


#Visualization 3: Histogram - Debt-to-Income Ratio Distribution
plt.figure(figsize=(10, 6))
ax = sb.histplot(
    data=df_debt_income_ratio,
    x='Debt-to-Income Ratio',
    bins='fd',
    kde=True,
    stat='percent',
    color='steelblue',
    edgecolor='black',
    kde_kws={'bw_adjust': 0.5}
)

ax.yaxis.set_major_formatter(ticker.PercentFormatter(xmax=100)) 
ax.xaxis.set_major_formatter(ticker.PercentFormatter(xmax=100)) 

plt.title('Distribution of Debt-to-Income Ratios Across Businesses', fontsize=14)
plt.xlabel('Debt-to-Income Ratio (%)', fontsize=12)
plt.ylabel('Share of businesses (%)', fontsize=12)

plt.tight_layout()
plt.savefig("visualization3.png", dpi=300, bbox_inches='tight')
plt.show()

#Visualization 4: Stacked Bar Chart - Total Liabilities and Total Equity by Business
df_viz4 = df_input[['Business ID', 'Total Liabilities', 'Total Equity']]
df_viz4 = df_viz4.sort_values(by='Total Liabilities', ascending=False).head(20)

#Set index to Business ID
df_viz4.set_index('Business ID', inplace=True)

ax = df_viz4.plot(
    kind='bar',
    stacked=True,
    figsize=(16, 6),
    color=['#1f77b4', '#ff7f0e'],
    edgecolor='black'
)

# Formatting
ax.set_title('Top 20 Businesses by Total Liabilities (with Equity)', fontsize=14)
ax.set_xlabel('Business ID', fontsize=12)
ax.set_ylabel('Amount (USD)', fontsize=12)
ax.legend(title='Category')
plt.xticks(rotation=45, ha='right')
ax.yaxis.set_major_formatter(
    ticker.FuncFormatter(lambda x, pos: f'{x/1_000_000_000:.2f}B'))

plt.tight_layout()
plt.savefig("visualization4.png", dpi=300, bbox_inches='tight')
plt.show()

```
